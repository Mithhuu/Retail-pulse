Import libraries and load data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('/content/RetailPulse_Cleaned.csv')

df['Order Date'] = pd.to_datetime(df['Order Date'])

print("Dataset loaded successfully!")
```

Analyse Year-wise Sales

```python
df['Year'] = df['Order Date'].dt.year

yearly_sales = df.groupby('Year')['Sales'].sum()

print(yearly_sales)
```

Analyse Monthly Sales

```python
df['Year-Month'] = df['Order Date'].dt.to_period('M')

monthly_sales = df.groupby('Year-Month')['Sales'].sum()

print(monthly_sales)
```

Find the Highest Sales Year

```python
best_year = yearly_sales.idxmax()

highest_sales = yearly_sales.max()

print("Best Sales Year:", best_year)

print("Highest Yearly Sales:", highest_sales)
```

Find the Highest Sales Month

```python
best_month = monthly_sales.idxmax()

highest_month_sales = monthly_sales.max()

print("Best Sales Month:", best_month)

print("Highest Monthly Sales:", highest_month_sales)
```

Find the Lowest Sales Month

```python
lowest_month = monthly_sales.idxmin()

lowest_month_sales = monthly_sales.min()

print("Lowest Sales Month:", lowest_month)

print("Lowest Monthly Sales:", lowest_month_sales)
```

Calculate Yearly Sales Growth

```python
yearly_growth = yearly_sales.pct_change() * 100

print(yearly_growth)
```
