```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv('/content/RetailPulse_Cleaned.csv')

df['Order Date'] = pd.to_datetime(df['Order Date'])

print("Dataset loaded successfully!")
```

Display Key Performance Indicators (KPIs)

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_orders = df['Order ID'].nunique()
total_quantity = df['Quantity'].sum()

profit_margin = (total_profit / total_sales) * 100

print("========== RETAIL PULSE DASHBOARD ==========")

print("Total Sales: $", round(total_sales, 2))
print("Total Profit: $", round(total_profit, 2))
print("Total Orders:", total_orders)
print("Total Quantity Sold:", total_quantity)
print("Profit Margin:", round(profit_margin, 2), "%")
```

Create Dashboard Charts
**Chart 1 – Sales by Category**

```python
category_sales = df.groupby('Category')['Sales'].sum()

category_sales.plot(kind='bar', title='Sales by Category')

plt.xlabel('Category')
plt.ylabel('Sales')
plt.show()
```

**Chart 2 – Profit by Region**

```python
region_profit = df.groupby('Region')['Profit'].sum()

region_profit.plot(kind='bar', title='Profit by Region')

plt.xlabel('Region')
plt.ylabel('Profit')
plt.show()
```

**Chart 3 – Monthly Sales Trend**

```python
df['Year-Month'] = df['Order Date'].dt.to_period('M')

monthly_sales = df.groupby('Year-Month')['Sales'].sum()

monthly_sales.plot(kind='line', figsize=(12,5), title='Monthly Sales Trend')

plt.xlabel('Month')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.show()
```

**Chart 4 – Profit by Subcategory**

```python
subcategory_profit = df.groupby('Sub-Category')['Profit'].sum()

subcategory_profit.sort_values().plot(
    kind='bar',
    figsize=(12,5),
    title='Profit by Subcategory'
)

plt.xlabel('Subcategory')
plt.ylabel('Profit')
plt.xticks(rotation=90)
plt.show()
```
