```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('/content/RetailPulse_Cleaned.csv')

df['Order Date'] = pd.to_datetime(df['Order Date'])

print("Dataset loaded successfully!")
df.head()
```

Sales distribution by category (Pie Chart)

```python
category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(7,7))

plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Sales Distribution by Category')
plt.savefig('RetailPulse_Sales_Distribution.png',
            dpi=300, bbox_inches='tight')
plt.show()
```

Profit by category (Bar Chart)

```python
category_profit = df.groupby('Category')['Profit'].sum()

plt.figure(figsize=(8,5))

plt.bar(category_profit.index, category_profit.values)

plt.title('Profit by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Profit')
plt.savefig('RetailPulse_Profit_Chart.png',
            dpi=300, bbox_inches='tight')
plt.show()
```

Sales comparison by region

```python
region_sales = df.groupby('Region')['Sales'].sum()

plt.figure(figsize=(8,5))

plt.bar(region_sales.index, region_sales.values)

plt.title('Sales Comparison by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.savefig('RetailPulse_Regional_Sales.png',
            dpi=300, bbox_inches='tight')
plt.show()
```

Monthly sales trend (Line Chart)

```python
df['Year-Month'] = df['Order Date'].dt.to_period('M')

monthly_sales = df.groupby('Year-Month')['Sales'].sum()

plt.figure(figsize=(14,6))

plt.plot(monthly_sales.index.astype(str), monthly_sales.values)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig('RetailPulse_Monthly_Trend.png',
            dpi=300, bbox_inches='tight')
plt.show()
```

Sales vs. Profit (Scatter Plot)

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Sales'], df['Profit'])

plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.savefig('RetailPulse_Profit_Chart.png',
            dpi=300, bbox_inches='tight')
plt.show()
```
